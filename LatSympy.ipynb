```python
import sympy as sp

x = sp.symbols('x')
f = 5 * (x**4) / 2
print(f)
```

```python
sp.pprint(f) #Pretty print (Memperbagus tampilan)
```

```python
fDiff = sp.diff(f, x)
print(fDiff, end="\n\n\n")
sp.pprint(fDiff)
```

#FUNCTION f(x)

```python
value_at_4 = fDiff.subs(x, 4)
print(value_at_4)

func = sp.lambdify(x, fDiff)
print(func(4))
```

#LATEX - SYMPY - LATEX

```python
import sympy as sp
from sympy.parsing.latex import parse_latex
import matplotlib.pyplot as plt

def latex_to_sympy(latex_code):
    return parse_latex(latex_code)

def process_expression(expression):
    x = sp.symbols('x')
    processed_expr = sp.diff(expression, x)
    return processed_expr

def sympy_to_latex(expression):
    return sp.latex(expression)

def display_latex(input_latex, output_latex):
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 6))
    fig.suptitle("Terjemahan fungsi yang akan diturunkan:", fontsize=16)

    ax1.text(0.5, 0.5, f"${input_latex}$", fontsize=20, ha='center', va='center')

    ax2.text(0.5, 0.5, f"${output_latex}$", fontsize=20, ha='center', va='center')
    ax2.set_title("Terjemahan Hasil Turunan:", fontsize=14)

    plt.tight_layout()
    plt.show()

def main():
    latex_input = input("Masukkan fungsi untuk diturunkan:")

    sympy_expr = latex_to_sympy(latex_input)
    processed_expr = process_expression(sympy_expr)
    latex_output = sympy_to_latex(processed_expr)
    print("Hasil penurunan (Latex):", latex_output)

    display_latex(latex_input, latex_output)

if __name__ == "__main__":
    main()
```
