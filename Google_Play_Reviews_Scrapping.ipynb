```python
from google_play_scraper import reviews, Sort
import pandas as pd

review_data, _ = reviews(
    "com.supercell.clashroyale",
    lang="id",
    country="id",
    count=2000,
    sort=Sort.NEWEST
)

df = pd.DataFrame(review_data)
df.to_csv("clash_royale_reviews.csv", index=False)
```
