```python
from langdetect import detect

# Encrypted sentence
encrypted_sentence = "Dqdn#gld#dgd#olpd"

def caesar_cipher_decrypt(text, shift):
    decrypted = []
    for char in text:
        decrypted.append(chr((ord(char) - shift) % 256))
    return ''.join(decrypted)

def brute_force_decrypt(encrypted_text):
    for shift in range(1, 256):
        decrypted_text = caesar_cipher_decrypt(encrypted_text, shift)
        print(f"Shift {shift}: {decrypted_text}")

        try:
            lang = detect(decrypted_text)
            if lang == 'en' or lang == 'id':
                print(f"\nPossible decryption found with shift {shift}: {decrypted_text}")
                break
        except:
            continue

# Run brute force decryption
brute_force_decrypt(encrypted_sentence)
```
