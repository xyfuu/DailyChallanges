#1

```python
import sympy as sp
from IPython.display import display, Math

R = 1.1 * 10 * sp.exp(-6) * 1.5 / (0.5 * sp.exp(-6))
lR = sp.latex(R)
display(Math(f"1. {lR}"))
```

#2

```python
Rx = 50 * sp.cos(sp.pi/4) + 60
Ry = 50 * sp.sin(sp.pi/4)

Res = (Rx**2 + Ry**2)**(1/2)
lRes = sp.latex(Res)
display(Math(f"2A. {lRes}"))

deg = sp.atan(Ry/Rx)
ldeg = sp.latex(deg)
display(Math(f"2B. {ldeg}"))
```

#3

```python
x = sp.symbols("x")

a = sp.sqrt(x+sp.sqrt(x+sp.sqrt(x)))
aDiff = sp.diff(a, x)
laDiff = sp.latex(aDiff)
display(Math(f"3A. {laDiff}"))

b = (5*x/sp.Pow(1-x, 2/3)) + sp.Pow(sp.cos(2*x+1), 2)
bDiff = sp.diff(b, x)
lbDiff = sp.latex(bDiff)
display(Math(f"3B. {lbDiff}"))

c = sp.tan(sp.sqrt((1-sp.cos(x))/(1+sp.cos(x))))
cDiff = sp.diff(c, x)
lcDiff = sp.latex(cDiff)
display(Math(f"3C. {lcDiff}"))
```

#4A

```python
import numpy as np

q = np.array([(7*4, 7*3),
              (7*6, 7*7)])
w = np.array([(2*1, 2*0),
              (2*5, 2*3)])

res = q-w
lres = sp.latex(sp.Matrix(res))
display(Math(f"4A. {lres}"))
```

#4B

```python

o = np.array([(-3*1, -3*sp.sqrt(5)),
              (-3*sp.sqrt(5), -3*9)])
p = np.array([(sp.sqrt(5), 0),
              (-2, sp.sqrt(5))])

res = o.dot(p)
lres = sp.latex(sp.Matrix(res))
display(Math(f"4B. {lres}"))
```

#4C

```python

h = np.array([[2, 3, 1],
              [2, -5, 1],
              [3, -2, 2]])

j = np.array([11, -5, 5])

res = np.linalg.solve(h, j)
lres = sp.latex(sp.Matrix(res))
display(Math(f"4C. {lres}"))
```
