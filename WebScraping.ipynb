#100 Pages Review + CSV

```python
import csv
collection = []

for page in range(1, 101):
  url = f"https://letterboxd.com/film/queen-of-tears-2024/reviews/by/activity/page/{page}/"
  response = requests.get(url)
  soup = BeautifulSoup(response.content, 'html.parser')
  reviews = soup.find_all('li', class_='film-detail')

  for review in reviews:
      username = review.find('strong', class_='name').getText()
      text = review.find('div', class_='body-text').getText()

      collection.append([username, text])
      print(username, text)

def export_to_csv(list, header, to=''):
  with open(to, mode='w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerow(header)
    writer.writerows(list)

export_to_csv(collection, ['username', 'text'], '100-page-reviews.csv')
```

#All Pages

```python
collection = []
page = 1

while True:
    url = f"https://letterboxd.com/film/queen-of-tears-2024/reviews/by/activity/page/{page}/"
    response = requests.get(url)
    soup = BeautifulSoup(response.content, 'html.parser')
    reviews = soup.find_all('li', class_='film-detail')

    if not reviews:  # If no reviews are found, break the loop
        break

    for review in reviews:
        username = review.find('strong', class_='name').getText()
        liked = review.find('span', class_='has-icon icon-16 icon-liked')
        like = "Liked the movie" if liked else 'No like given'
        rating_element = review.find('span', class_='rating')
        rating = rating_element.getText() if rating_element else 'No Rating'

        if rating != 'No Rating':
            num_stars = rating.count('★')
            if '½' in rating:
                num_stars += 0.5
            rating = num_stars
        date = review.find('span', class_='_nobr').getText()
        text = review.find('div', class_='body-text').getText()

        collection.append([username, like, rating, date, text])
        print(username, like, rating, date, text)

    page += 1
```

```python
import csv

def export_to_csv(list, header, to=''):
  with open(to, mode='w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerow(header)
    writer.writerows(list)

export_to_csv(collection, ['username', 'like', 'rating', 'date', 'text'], 'all-page-reviews.csv')
```
