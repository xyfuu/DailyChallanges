```python
import requests
from bs4 import BeautifulSoup
```

```python
title = input().lower()
title = title.replace(" ", "-")
episode = input()
url = f"https://oploverz.my/{title}-episode-{episode}-subtitle-indonesia/"
response = requests.get(url)
soup = BeautifulSoup(response.content, 'html.parser')

status = False
links = soup.find_all('a', class_=False)
for link in links:
    href = link.get('href')
    if href and 'mediafire.com' in href:
        print(href)
        status = True

if status == False:
  print("No download links provided")
```
