#Graph in Python using Graphvis

#### Alice -> Alice x Eric, Alice x Fred
#### Bob -> Bob x Cody
#### Cody -> -
#### Dan -> -
#### Eric -> -
#### Fred -> Fred x Bob, Fred x Eric

G = (v, E)
  = ({A, B, C, D, E, F}, {A, E}, {A, F}, {B, C}, {F, B}, {F, E})

```python
from graphviz import Graph

nodes = {'Alice', 'Bob', 'Cody', 'Dan', 'Eric', 'Fred'}
edges = {('Alice', 'Eric'), ('Alice', 'Fred'), ('Bob', 'Cody'), ('Fred', 'Bob'), ('Fred', 'Eric')}

g = Graph('Handshakes')

for node in nodes:
  g.node(node)

for edge in edges:
  g.edge(*edge)

g.render('handshakes', view=True)
```

```python
from IPython.display import display, Image

g.render('handshakes', format='png')
display(Image(filename='handshakes.png'))
```

# Graph Berarah

```python
from graphviz import Digraph

dot = Digraph(comment='Handshakes')

nodes = {'Alice', 'Bob', 'Cody', 'Dan', 'Eric', 'Fred'}
for node in nodes:
  dot.node(node)

dot.edge('Alice', 'Eric')
dot.edge('Alice', 'Fred')

dot.edge('Bob', 'Cody')

dot.edge('Fred', 'Bob')
dot.edge('Fred', 'Eric')

image_bytes = dot.pipe(format='png')
display(Image(image_bytes))
```

# Graph Berbobot

```python
dot = Graph(comment='Cities and Distances')

cities = ['A', 'B', 'C', 'D', 'E']
for city in cities:
  dot.node(city)

distances = {
    ('A', 'B'): 5,
    ('A', 'C'): 10,
    ('B', 'C'): 3,
    ('B', 'D'): 2,
    ('C', 'D'): 7,
    ('C', 'E'): 1,
    ('D', 'E'): 4,
}

for (start, end), distance in distances.items():
  dot.edge(start, end, label=str(distance))

image_bytes = dot.pipe(format='png')
display(Image(image_bytes))
```

#Find shortest path (A->C)

```python
import heapq

# Define the graph as a dictionary where each city has a list of connected cities and distances
graph = {
    'A': {'B': 5, 'C': 10},
    'B': {'A': 5, 'C': 3, 'D': 2},
    'C': {'A': 10, 'B': 3, 'D': 7, 'E': 1},
    'D': {'B': 2, 'C': 7, 'E': 4},
    'E': {'C': 1, 'D': 4},
}

# Dijkstra's algorithm function
def dijkstra(graph, start, goal):
    queue = [(0, start, [])]  # (cost, node, path)
    visited = set()

    while queue:
        (cost, node, path) = heapq.heappop(queue)
        if node in visited:
            continue

        path = path + [node]
        visited.add(node)

        if node == goal:
            return cost, path

        for next_node, weight in graph[node].items():
            if next_node not in visited:
                heapq.heappush(queue, (cost + weight, next_node, path))

    return float("inf"), []  # Return infinity if there is no path

# Find the shortest path from A to C
shortest_distance, shortest_path = dijkstra(graph, 'A', 'E')
print(f"The shortest path from A to E is: {shortest_path} with a distance of {shortest_distance}")
```
